# Handling the missing values

```python
df['last review'] = pd.to_datetime(df['last review'], errors = 'coerce')
df.info()
```

```python
df.fillna({'reviews per month' : 0, 'last review': df['last review'].min()},inplace = True)
```

```python
df.dropna(subset = ['NAME','host name'],inplace = True)
```

```python
print(df.isnull().sum())
```

```python
 df = df.drop(columns = ['license','house_rules'], errors = 'ignore')
df
```

```python
# Remove dollar signs and convert to float
df['price']=df['price'].replace('[\$,]','',regex = True).astype(float)
df['service fee'] = df['service fee'].replace('[\$,]','',regex=True).astype(float)
```

```python
df.head()
```

# Revomes duplicates

```python
df.drop_duplicates(inplace = True)
```

```python
df.info()
```

# visualization
# Q.1 -- What is the distribution of listing prices?

```python
plt.figure(figsize = (10,6))
sms.histplot(df['price'],bins = 50,kde = True,color = 'red')
plt.title('distrbution of listing price')
plt.xlabel("price $")
plt.ylabel("frequency")
plt.show()
```

```python
The histogram shows a fairly even distribution of
listing prices across different prices ranges,indicating no
particular concentration of listings in any specficed price range.
THE KDE line helps visulazation this even spread more clerly,
confirming that the dataset contains listings with a wide variety of prices
```

# Q.2---How are different room types distributed?

```python
df['room type']
```

```python
plt.figure(figsize = (8,5))
sms.countplot(x = 'room type',data = df,color = 'hotpink')
plt.title("Room type of distribution")
plt.xlabel("room types")
plt.ylabel("count")
plt.show()
```

# How are listings distributed across different neighborhoods?

```python
plt.figure(figsize = (12,8))
sms.countplot(y = "neighbourhood group",data = df,order = df['neighbourhood group'].value_counts().index)
plt.title('Number of Listing by Neighbourhood Group')
plt.xlabel('count')
plt.ylabel('neighbourhood group')
plt.show()
```

# what is the realtionship between price and room types ?

```python
plt.figure(figsize = (10,6))
sms.boxplot(x = "room type",y= 'price',hue='room type',data = df, palette ='Set1')
plt.title('Price vs Room Type',frontsize=16)
plt.xlabel('Room Type')
plt.ylabel('Price ($)')
plt.legend(title = 'Room type')
plt.tight_layout()
plt.show()
```

# how has the number of review change over with times

```python
df ['last review'] = pd.to_datetime(df['last review'])
reviews_over_time = df.groupby(df['last review'].dt.to_period('M')).size()

plt.figure(figsize = (12,6))
reviews_over_time.plot(kind ='line',color = 'red')
plt.title('Number of review over Time')
plt.xlabel("Date")
plt.ylabel("Number of reviews")
plt.show()
```
